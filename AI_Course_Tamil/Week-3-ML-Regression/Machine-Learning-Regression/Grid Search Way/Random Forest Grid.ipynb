```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
dataset
```

```python
# Dataset has 1338 Rows X 6 Columns
# We have to convert the following nominal fields to numerical using one-hot encoding.
# The nominal fields are smoker, sex.
```

```python
# Preprocessing
dataset = pd.get_dummies(dataset, dtype="int64", drop_first=True)
dataset
```

```python
#  Display Columns
dataset.columns
```

```python
# Identify independent variables
independent = dataset[["age", "bmi", "children", "sex_male", "smoker_yes"]]
independent
```

```python
# Identify dependent variable
dependent = dataset[["charges"]]
dependent
```

```python
# Import GridSearchCV to test multiple hyperparameter combinations
# using cross-validation and find the best combination
from sklearn.model_selection import GridSearchCV

# Import RandomForestRegressor for Random Forest Regression
from sklearn.ensemble import RandomForestRegressor

# Define the hyperparameters and values that GridSearchCV should try
param_grid = {
    # Define the criterion used to measure the quality of each split
    # squared_error -> minimizes squared errors
    # absolute_error -> minimizes absolute errors
    'criterion': ['squared_error', 'absolute_error'],

    # Define the number of features considered when finding the best split
    # None -> consider all features
    # sqrt -> consider square root of the total number of features
    # log2 -> consider log base 2 of the total number of features
    'max_features': [None, 'sqrt', 'log2'],

    # Define the number of decision trees in the Random Forest
    'n_estimators': [10, 100]
}

# Create the GridSearchCV object
# RandomForestRegressor() -> the machine learning model we want to tune
# param_grid -> contains all hyperparameter combinations
# refit=True -> retrains the best model using the complete data after tuning
# verbose=3 -> displays detailed progress while GridSearchCV is running
# n_jobs=-1 -> uses all available CPU cores to speed up the process
grid = GridSearchCV(
    RandomForestRegressor(),
    param_grid,
    refit=True,
    verbose=3,
    n_jobs=-1
)

# Start the hyperparameter search
# GridSearchCV will try every possible combination of the parameters
# independent -> input/features
# dependent -> target/output
grid.fit(independent, dependent)
```

```python
# Get all results generated by GridSearchCV
# cv_results_ contains information about every hyperparameter combination tested,
# including parameters, training scores, validation scores, fit time, and ranking
re = grid.cv_results_

# Convert the GridSearchCV results dictionary into a pandas DataFrame
# This makes the results easier to read, filter, sort, and analyze
table = pd.DataFrame.from_dict(re)

# Display the complete results table
table
```
